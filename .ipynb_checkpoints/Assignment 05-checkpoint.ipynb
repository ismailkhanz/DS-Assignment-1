# Question 1

Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.

Example: 'Welcome'
Output: 3 vowels

```python
def count_vowels(text):
    vowels = "aeiouAeiou"
    count = 0
    for i in text:
        if i in vowels:
            count += 1
    return count
result = count_vowels("Welcome")
print(result,"vowels")
```

# Question 3

Q3. Write a Python function display_prices that takes a product name and an arbitrary number of keyword arguments representing product prices. The function should display each product and its price.

Expected Output:

Product: Laptop

Amazon: $800

eBay: $750

Walmart: $780

```python
def display_prices(product,**prices):
    print("Product:",product)
    for company,price in prices.items():
        print(f"{company}:${price}")
display_prices("Laptop",Amazon=800,eBay=750,Walmart=780)
```

# Question 6

Q6.

Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:

- calculate_area: This function should calculate the area of the rectangle.
- calculate_perimeter: This function should calculate the perimeter of the rectangle.
- The main function should return both the area and the perimeter of the rectangle.

Expected Output:

Area: 15

Perimeter: 16

```python
def calculate_rectangle(lenght,widht):
    def calculate_area():
        return lenght * widht
    def calculate_perimeter():
        return 2*(lenght + widht)
    return calculate_area(),calculate_perimeter()
area,perimeter = calculate_rectangle(5,3)
print("Area:",area)
print("Perimeter:",perimeter)
```

# Question 8

Q8. Write a nested function to calculate the total hotel bill based on the number of nights. A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

Expected Output: Room Cost: 15000 Tax: 1500.0 Total Bill: 16500.0

```python
def total_bill(nights):
    
    def hotel_charges():
        return nights * 5000  
    def tax():
        return hotel_charges() * 0.10
    def total():
        return hotel_charges() + tax()
    print("Room Cost:",hotel_charges(),"Tax",tax(),"Total Bill:",total())
total_bill(3)
```

Q9.User Login Validation A website checks a user's username and password before allowing access. Write a nested function to validate the login details

```python
def login(username,password):
    def user():
        return username == "Ismailkhnaz"
    def passwordd():
        return password == 12345
    if user() and passwordd():
        return "Login Successful"
    elif not user():
        return "Invalid Username"
    else:
        return "Invalid Password"
print(login(username= "Ismailkhnaz",password=12345))
print(login(username= "Ismailkhan",password=12345))
print(login(username= "Ismailkhnaz",password=124))
```

# Question 10

Q10.

Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

An inner function calculate_total() → multiplies price * quantity.
An inner function apply_discount(total) → subtracts discount from total (but never below 0).
The outer function should return the final bill amount.

Expected Output:

Final Bill = 600 PKR (no discount)
Final Bill = 500 PKR (with discount)

```python
def shopping_bill(price,quantity,discount=0):
    def calculate_total():
        return price * quantity
    def apply_discount(total):
        return max (total - discount,0)
    total = calculate_total()
    return apply_discount(total)
print("Final Bill=",shopping_bill(10000,3),"PKR")
print("Final Bill=",shopping_bill(5000,3,discount=0),"PKR")
```
